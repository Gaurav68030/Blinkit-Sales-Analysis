```python
total_sales = round(df['Sales'].sum(),2)
print('Total Sales: 'f"₹{total_sales:,.0f}")
```

```python
avg_sales = round(df['Sales'].mean())
print('avg_sales: 'f"₹{avg_sales:,.0f}")
```

```python
avg_rating = round(df['Rating'].mean(),2)
print(f'Average Rating:' ,avg_rating)
```

```python
no_of_items_sold = df['Sales'].count()
print(f"No of Items Sold: {no_of_items_sold:,.0f}")
```

```python
sales_by_item = df.groupby('Item Type')['Sales'].sum()

plt.figure(figsize=(10, 6))
plt.bar(sales_by_item.index, sales_by_item.values)

plt.title('Total Sales by Item Type')
plt.xlabel('Item Type')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')

plt.show()
```

```python
Sales_by_outlet = df.groupby('Outlet Type')['Sales'].sum()
plt.figure(figsize=(8, 5))
plt.bar(Sales_by_outlet.index, Sales_by_outlet.values)
plt.title('Total Sales by Outlet Type')
plt.xlabel('Outlet Type')
plt.ylabel('Total Sales')
plt.show()



```

```python
sales_by_size = df.groupby('Outlet Size')['Sales'].sum()
plt.figure(figsize=(8,5))
plt.bar(sales_by_size.index,sales_by_size.values)
plt.title('Total Sales by Outlet Size')
plt.xlabel('Outlet Size')
plt.ylabel('Total Sales')
plt.show()
```

```python
sales_by_location = df.groupby('Outlet Location Type')['Sales'].sum()
plt.figure(figsize=(8, 5))
plt.bar(sales_by_location.index, sales_by_location.values)

plt.title('Total Sales by Outlet Location Type')
plt.xlabel('Outlet Location Type')
plt.ylabel('Total Sales')

plt.show()
```

```python
sales_by_year = df.groupby('Outlet Establishment Year')['Sales'].sum().sort_index()

plt.figure(figsize=(10, 5))
plt.plot(sales_by_year.index, sales_by_year.values, marker='o')

plt.title('Total Sales by Outlet Establishment Year')
plt.xlabel('Outlet Establishment Year')
plt.ylabel('Total Sales')

plt.show()
```

## Final Business Insights

#### Which item type has the highest sales?

```python
df.groupby('Item Type')['Sales'].sum().idxmax()
```

#### Which outlet type performs best?

```python
df.groupby('Outlet Type')['Sales'].sum().idxmax()
```

#### Which outlet size generates the most sales?

```python
df.groupby('Outlet Size')['Sales'].sum().idxmax()
```

#### Which location type performs best?

```python
df.groupby('Outlet Location Type')['Sales'].sum().idxmax()
```

#### How have sales changed by establishment year?

```python
df.groupby('Outlet Establishment Year')['Sales'].sum().sort_index()
```

#### Is Item Visibility strongly related to Sales?


```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['Item Visibility'],
    df['Sales'],
    alpha=0.5
)
plt.title('Item Visibility vs Sales')
plt.xlabel('Item Visibility')
plt.ylabel('Sales')

plt.show()
```

#### Is Rating strongly related to Sales?

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['Rating'],
    df['Sales'],
    alpha=0.5
)

plt.title('Rating vs Sales')
plt.xlabel('Rating')
plt.ylabel('Sales')

plt.show()
```

#### Which Fat-Content category contributes more to Sales?

```python
sales_by_fat = df.groupby('Item Fat Content')['Sales'].sum()

plt.figure(figsize=(8, 5))
plt.bar(sales_by_fat.index, sales_by_fat.values)

plt.title('Total Sales by Fat Content')
plt.xlabel('Item Fat Content')
plt.ylabel('Total Sales')

plt.show()
```
